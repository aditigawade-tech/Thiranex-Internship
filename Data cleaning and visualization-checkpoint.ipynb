## Handling outliers

```python
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

```python
df.drop(columns=['Unnamed: 0'], axis=1, inplace=True, errors='ignore')
```

```python
df.info()
```

```python
Q1=df["price"].quantile(0.25)
Q3=df["price"].quantile(0.75)
IQR = Q3-Q1
lower_limit=Q1-1.5*IQR
upper_limit=Q3+1.5*IQR
print("Lower_Limit:",lower_limit)
print("Upper_Limit:",upper_limit) 
```

## Use capping method

```python
df["price"]=df["price"].clip(
    lower=0,
    upper=99128 )
```

```python
sns.boxplot(x=df["price"])
plt.show()
```

```python
# checking how many prices were affected 
(df["price"]>99128).sum()
```

## Data Visualization 

```python
df.info()
```

```python
#Flight Price Distribution
plt.figure(figsize=(8,5))

sns.histplot(
    df["price"],
    bins=50,
    kde=True
)

plt.title("Flight Price Distribution")
plt.xlabel("Price")
plt.show()
```

```python

#Average Price by Airline
plt.figure(figsize=(10,5))

sns.barplot(
    x="airline",
    y="price",
    data=df
)

plt.xticks(rotation=45)
plt.title("Average Flight Price by Airline")
plt.show()
```

```python
#Class vs Price
plt.figure(figsize=(6,4))

sns.boxplot(
    x="class",
    y="price",
    data=df
)

plt.title("Price Variation by Class")
plt.show()
```

```python
#Number of Flights by Airline
plt.figure(figsize=(10,5))

sns.countplot(
    x="airline",
    data=df
)

plt.xticks(rotation=45)
plt.title("Number of Flights by Airline")
plt.show()
```

```python
#Duration vs Price Relationship
plt.figure(figsize=(8,5))

sns.scatterplot(
    x="duration",
    y="price",
    data=df
)

plt.title("Flight Duration vs Price")
plt.show()
```

```python
#Correlation Heatmap
plt.figure(figsize=(6,4))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

## Insights

- Flight prices vary significantly depending on airline and ticket class.
- Business class tickets generally have higher prices than Economy class.
- Flight duration and days left before departure influence pricing patterns.
- Different airlines show different average pricing trends.
